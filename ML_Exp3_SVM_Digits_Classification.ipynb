# ML Experiment 3: Support Vector Machine on Digits Dataset
**Aim:** Classify handwritten digits (sklearn `load_digits`) with an SVM.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import svm
from sklearn.datasets import load_digits
from sklearn.manifold import Isomap
from sklearn.metrics import confusion_matrix, classification_report

digits = load_digits(n_class=10)
print(digits.data.shape)
```

```python
digits['data'][0].reshape(8, 8)
```

## Sample images

```python
fig, axes = plt.subplots(8, 8, figsize=(8, 8), subplot_kw={'xticks':[], 'yticks':[]})
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='binary', interpolation='nearest')
    ax.text(0.05, 0.05, str(digits.target[i]), transform=ax.transAxes, color='green')
plt.show()
```

## Isomap 2-D projection

```python
iso = Isomap(n_components=2)
proj = iso.fit_transform(digits.data)
plt.figure(figsize=(8,6))
plt.scatter(proj[:, 0], proj[:, 1], c=digits.target, alpha=0.5, cmap='Spectral')
plt.colorbar(ticks=range(10), label='Digits Value')
plt.clim(-0.5, 9.5)
plt.show()
```

```python
def view_digit(index):
    plt.imshow(digits.images[index], cmap=plt.cm.gray_r)
    plt.title('Digit label: {}'.format(digits.target[index]))
    plt.show()
view_digit(5)
```

## Train SVM (first 1500 samples) and test on the rest

```python
main_data = digits['data']
targets = digits['target']
clf = svm.SVC(gamma=0.001, C=100)
clf.fit(main_data[:1500], targets[:1500])
predictions = clf.predict(main_data[1501:])
print(list(zip(predictions, targets[1501:]))[:20])
```

```python
cm = confusion_matrix(targets[1501:], predictions)
plt.figure(figsize=(8,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu')
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.show()
print(classification_report(targets[1501:], predictions))
```
