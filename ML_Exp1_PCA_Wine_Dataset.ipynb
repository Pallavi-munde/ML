# ML Experiment 1: Principal Component Analysis (PCA) on Wine Dataset
**Aim:** Reduce the wine-quality features to 2 principal components and visualise them (red vs white wine and quality).

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Load data (UCI Wine Quality: red + white)
If the download fails, upload `winequality-red.csv` (and optionally `winequality-white.csv`).

```python
base = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/"
try:
    red = pd.read_csv(base + "winequality-red.csv", sep=";")
    white = pd.read_csv(base + "winequality-white.csv", sep=";")
except Exception as e:
    print("Download failed:", e, "\nPlease upload the CSV file(s).")
    from google.colab import files
    up = files.upload()
    red = pd.read_csv("winequality-red.csv", sep=";")
    white = pd.read_csv("winequality-white.csv", sep=";") if "winequality-white.csv" in up else red.iloc[0:0]
red["type"] = "red"; white["type"] = "white"
data = pd.concat([red, white], ignore_index=True)
print(data.shape)
data.head()
```

## Separate features and standardise

```python
X = data.drop(["quality", "type"], axis=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Features:", X.shape[1])
```

## Apply PCA (2 components)

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance:", pca.explained_variance_ratio_.sum())

pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Quality"] = data["quality"].values
pca_df["Type"] = data["type"].values
pca_df.head()
```

## Visualise: coloured by quality

```python
plt.figure(figsize=(8,6))
sc = plt.scatter(pca_df["PC1"], pca_df["PC2"], c=pca_df["Quality"], cmap="viridis", alpha=0.7)
plt.colorbar(sc, label="Wine Quality")
plt.xlabel("Principal Component 1"); plt.ylabel("Principal Component 2")
plt.title("PCA of Wine Dataset (coloured by quality)")
plt.show()
```

## Visualise: red vs white wine

```python
plt.figure(figsize=(8,6))
for t, c in [("red","crimson"),("white","goldenrod")]:
    s = pca_df[pca_df.Type==t]
    if len(s): plt.scatter(s.PC1, s.PC2, c=c, alpha=0.5, label=t, s=12)
plt.legend(); plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("PCA: Red vs White wine")
plt.show()
```

```python
print("Original features:", X.shape[1])
print("Reduced features :", 2)
print(f"Variance preserved: {pca.explained_variance_ratio_.sum()*100:.2f} %")
```
