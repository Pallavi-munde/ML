# ML Experiment 4: K-Means Clustering on Iris Dataset
**Aim:** Cluster iris flowers with K-Means and choose k with the elbow method.

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
plt.style.use('seaborn-v0_8')
```

## Load Iris data (same columns as iris.csv)

```python
from sklearn.datasets import load_iris
iris = load_iris()
data = pd.DataFrame(iris.data, columns=['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm'])
data.insert(0, 'Id', range(1, len(data)+1))
data['Species'] = [ 'Iris-' + iris.target_names[i] for i in iris.target ]
# To use your own file instead:  data = pd.read_csv('iris.csv')
data.head()
```

```python
print(data.tail()); print(len(data)); print(data.shape); print(data.columns)
print(data.dtypes); data.info(); print(data.describe()); print(data.isnull().sum())
```

```python
data.drop('Id', axis=1, inplace=True)
print(data['Species'].value_counts())
target_data = data.iloc[:, 4]
clustering_data = data.iloc[:, [0,1,2,3]]
```

```python
plt.figure(figsize=(8,5))
plt.scatter(data['SepalLengthCm'], data['SepalWidthCm'], color='#f73434')
plt.xlabel('Sepal Length (cm)'); plt.ylabel('Sepal Width (cm)'); plt.title('Sepal length vs width')
plt.show()
```

## Elbow method

```python
inertia = []
for i in range(1, 11):
    inertia.append(KMeans(n_clusters=i, n_init=10, random_state=0).fit(clustering_data).inertia_)
plt.figure(figsize=(8,5))
plt.plot(range(1, 11), inertia, marker='o')
plt.axvline(x=3, color='r', linestyle='--')
plt.xlabel('Number of clusters'); plt.ylabel('Inertia (WCSS)'); plt.title('Elbow method')
plt.show()
print(inertia)
```

## Fit K-Means with k = 3

```python
kms = KMeans(n_clusters=3, init='k-means++', n_init=10, random_state=0)
clusters = clustering_data.copy()
clusters['Cluster_Prediction'] = kms.fit_predict(clustering_data)
print(clusters.head())
print(kms.cluster_centers_)
```

```python
colors = ['teal', 'lime', 'magenta']
plt.figure(figsize=(8,6))
for c in range(3):
    s = clusters[clusters.Cluster_Prediction == c]
    plt.scatter(s['SepalLengthCm'], s['SepalWidthCm'], c=colors[c], label=f'Cluster {c}')
plt.scatter(kms.cluster_centers_[:,0], kms.cluster_centers_[:,1], s=170, c='yellow', edgecolors='k', label='Centroids')
plt.xlim(4, 8); plt.ylim(1.8, 4.5)
plt.xlabel('Sepal Length (cm)'); plt.ylabel('Sepal Width (cm)'); plt.legend(); plt.show()
```
